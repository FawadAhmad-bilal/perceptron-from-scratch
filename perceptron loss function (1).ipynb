```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
import seaborn as sns
```

```python
x,y=make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,n_classes=2,
                        n_clusters_per_class=1,random_state=41
                        ,hypercube=False,class_sep=15)
```

```python
x.shape
```

```python
plt.scatter(x[:,0],x[:,1],c=y,cmap="winter",s=100)
plt.show()
```

```python

def perceptor(x,y): 
    w1=w2=b=1
    ir=0.01

    for j in range(1000):
        for i in range(x.shape[0]):
            
           z = w1*x[i][0] + w2*x[i][1]+b

           if z*y[i]>0:
             w1=w1+ir*y[i]*x[i][0]
             w2=w2+ir*y[i]*x[i][1]
             b=b+ir*y[i]
    return w1,w2,b
```

```python
w1,w2,b = perceptor(x,y)
```

```python
m=-(w1/w2)
c=-(b/w2)
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + c



plt.plot(x_input,y_input,color="red",linewidth=3)
plt.scatter(x[:,0],x[:,1],c=y,cmap="winter",s=100)
plt.ylim(-3,2)
```

```python
from sklearn.linear_model import Perceptron
```

```python
plf=Perceptron()
plf.fit(x,y)
print(plf.coef_,plf.intercept_)
```

```python
m,c
```

```python
print("Sklearn accuracy:", plf.score(x, y))
```
